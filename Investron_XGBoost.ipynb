```python
import numpy as np
import matplotlib.pyplot as plt
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

cols_to_use = [1,3,4,5]

# Load and prepare data
data = np.genfromtxt('ADANIPORTS.csv', delimiter=',', skip_header=1)
X = data[:, :-1]
Y = data[:, -1]

Y = np.where(Y > 0, 1, 0)


# Split data into train, validation, and test sets
cv_index = int(0.6 * len(X))
test_index = int(0.8 * len(X))

X_train, X_val, X_test = X[:cv_index], X[cv_index:test_index], X[test_index:]
Y_train, Y_val, Y_test = Y[:cv_index], Y[cv_index:test_index], Y[test_index:]

# Initialize and train XGBoost model
model = XGBClassifier(objective='binary:logistic', eval_metric='logloss', use_label_encoder=False)

model.fit(X_train, Y_train, eval_set=[(X_val, Y_val)], early_stopping_rounds=10, verbose=False)

# Evaluate model
Y_pred = model.predict(X_test)
test_accuracy = accuracy_score(Y_test, Y_pred)

print(f'Test Accuracy: {test_accuracy}')

# Plot training history if available (not applicable for XGBoost)
# Plotting code omitted as XGBoost does not have an epoch-based training history
```
