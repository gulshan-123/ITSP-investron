```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Input, Normalization, Dropout, LSTM, LeakyReLU
from keras.losses import MeanSquaredError
from keras.optimizers import Adam

import logging
logging.getLogger("tensorflow").setLevel(logging.ERROR)
tf.autograph.set_verbosity(0)

cols_to_use = [0,1,2,3]

nifty_symbols = [
    "ADANIPORTS", "ASIANPAINT", "AXISBANK", "BAJAJFINSV",
    "BAJFINANCE", "BPCL", "BRITANNIA", "CIPLA", "COALINDIA",
    "DIVISLAB", "DRREDDY", "EICHERMOT", "GRASIM", "HCLTECH",
    "HDFCBANK", "HDFCLIFE", "HEROMOTOCO", "HINDALCO", "HINDUNILVR",
    "ICICIBANK", "ICICIGI", "IOC", "INDUSINDBK", "INFY",
    "ITC", "JSWSTEEL", "KOTAKBANK", "LTTS", "LT",
     "MARICO", "MARUTI", "NESTLEIND"
]

# , "BAJAJ-AUTO" and "M&M", dont work

for symbol in nifty_symbols:
    data = np.genfromtxt(f'./data_hourly/{symbol}.csv', delimiter=',', skip_header=1)

    X = data[:,:-1]
    Y = data[:,-1]
    cv_index = int(0.6 * len(X))
    test_index = int(0.8 * len(X))
    print(len(X))

    X_train, X_val, X_test = X[:cv_index], X[cv_index:test_index], X[test_index:]
    Y_train, Y_val, Y_test = Y[:cv_index], Y[cv_index:test_index], Y[test_index:]


    X_train = X_train[:, cols_to_use]
    X_val = X_val[:, cols_to_use]
    X_test = X_test[:, cols_to_use]

    # Normalize the features
    norm_l = Normalization(axis=-1)
    norm_l.adapt(X_train)
    X_train_n = norm_l(X_train)
    X_val_n = norm_l(X_val)
    X_test_n = norm_l(X_test)

    # Reshape the data for LSTM
    X_train_n = tf.reshape(X_train_n, (X_train_n.shape[0], 1, X_train_n.shape[1]))
    X_val_n = tf.reshape(X_val_n, (X_val_n.shape[0], 1, X_val_n.shape[1]))
    X_test_n = tf.reshape(X_test_n, (X_test_n.shape[0], 1, X_test_n.shape[1]))


    tf.random.set_seed(42)
   # Check NaN or Inf in the input data
    print(f"NaN in X_train: {np.isnan(X_train_n).sum()}")  # Ensure no NaN in features
    print(f"Inf in X_train: {np.isinf(X_train_n).sum()}")  # Ensure no Inf in features
    print(f"NaN in Y_train: {np.isnan(Y_train).sum()}")    # Ensure no NaN in target
    print(f"Inf in Y_train: {np.isinf(Y_train).sum()}")    # Ensure no Inf in target

    # If any NaNs or Infs exist, replace them with 0 or some other value:
    # X_train_n = np.nan_to_num(X_train_n, nan=0.0)
    # X_val_n = np.nan_to_num(X_val_n, nan=0.0)
    # X_test_n = np.nan_to_num(X_test_n, nan=0.0)
    # Y_train = np.nan_to_num(Y_train, nan=0.0)
    # Y_val = np.nan_to_num(Y_val, nan=0.0)
    # Y_test = np.nan_to_num(Y_test, nan=0.0)



    # Define the LSTM model
    model = Sequential([
    Input(shape=(1, 4)),
    LSTM(128, activation='relu', name="LSTM_2", return_sequences=True),
    # Dropout(0.2),  # Add dropout to prevent overfitting
    LSTM(64, activation='relu', name="LSTM_1"),
    # Dropout(0.2),
    Dense(32, activation='relu'),
    # Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)  # Linear output (no activation)
])

 

    optimizer = Adam(learning_rate=0.0001, clipvalue=1.0)  # Clip gradients to 1.0
    model.compile(optimizer=optimizer, loss='mean_absolute_error')


    # Train the model
    history = model.fit(X_train_n, Y_train, validation_data=(X_val_n, Y_val), epochs=3, batch_size=32)

    # Plot the training and validation loss
    plt.plot(history.history['loss'], label='train_loss')
    plt.plot(history.history['val_loss'], label='val_loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.show()

    model.save(f'./models_hourly/{symbol}_model.keras')

    # Print the model summary
    model.summary()



# Evaluate the model on the test set
test_loss = model.evaluate(X_test_n, Y_test)
print(f'Test Loss: {test_loss}')

# # Predict on the test set
# Y_pred = model.predict(X_test_n)

# # Printing original and predicted values
# for i in range(len(Y_pred)):
#     print(f"Original: {Y_test[i]}, Predicted: {Y_pred[i][0]}")
```
